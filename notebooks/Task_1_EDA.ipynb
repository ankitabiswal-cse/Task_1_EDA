```python
# ============================================
# TASK 1 - EDA : LOAD SALES DATASET
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# --------------------------------------------
# 1. Find sales.csv automatically
# --------------------------------------------

possible_locations = [
    Path.home() / "OneDrive" / "Desktop",
    Path.home() / "Desktop",
    Path.home() / "OneDrive" / "Documents",
    Path.home() / "Documents"
]

csv_file = None

for location in possible_locations:
    if location.exists():
        matches = list(location.rglob("sales.csv"))
        if matches:
            csv_file = matches[0]
            break

# --------------------------------------------
# 2. Check whether file was found
# --------------------------------------------

if csv_file is None:
    print("❌ sales.csv file nahi mili.")
    print("\nPlease check ki file ka naam exactly 'sales.csv' hai.")
else:
    print("✅ File found successfully!")
    print("File location:", csv_file)

    # ----------------------------------------
    # 3. Load dataset
    # ----------------------------------------

    df = pd.read_csv(csv_file,sep="\t")

    # ----------------------------------------
    # 4. Basic cleaning
    # ----------------------------------------

    # Remove accidental unnamed columns
    df = df.loc[:, ~df.columns.str.contains("^Unnamed", case=False)]

    # Remove spaces from column names
    df.columns = df.columns.str.strip()

    # Convert Date column if available
    if "Date" in df.columns:
        df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

    # Convert numeric columns
    numeric_columns = ["Unit_Price", "Quantity", "Total_Sales"]

    for col in numeric_columns:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")

    # ----------------------------------------
    # 5. Dataset information
    # ----------------------------------------

    print("\n" + "=" * 50)
    print("DATASET LOADED SUCCESSFULLY")
    print("=" * 50)

    print("\nShape:", df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:", df.duplicated().sum())

    print("\nFirst 5 Rows:")
    display(df.head())

    # ----------------------------------------
    # 6. Statistical summary
    # ----------------------------------------

    print("\n" + "=" * 50)
    print("STATISTICAL SUMMARY")
    print("=" * 50)

    display(df.describe(include="all"))

    # ----------------------------------------
    # 7. Sales by Category
    # ----------------------------------------

    if "Category" in df.columns and "Total_Sales" in df.columns:

        print("\n" + "=" * 50)
        print("SALES BY CATEGORY")
        print("=" * 50)

        sales_by_category = (
            df.groupby("Category")["Total_Sales"]
            .sum()
            .sort_values(ascending=False)
        )

        display(sales_by_category)

        plt.figure(figsize=(8, 5))
        sales_by_category.plot(kind="bar")
        plt.title("Sales by Category")
        plt.xlabel("Category")
        plt.ylabel("Total Sales")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    # ----------------------------------------
    # 8. Sales by City
    # ----------------------------------------

    if "City" in df.columns and "Total_Sales" in df.columns:

        print("\n" + "=" * 50)
        print("SALES BY CITY")
        print("=" * 50)

        sales_by_city = (
            df.groupby("City")["Total_Sales"]
            .sum()
            .sort_values(ascending=False)
        )

        display(sales_by_city)

        plt.figure(figsize=(8, 5))
        sales_by_city.plot(kind="bar")
        plt.title("Sales by City")
        plt.xlabel("City")
        plt.ylabel("Total Sales")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    # ----------------------------------------
    # 9. Sales by Customer Type
    # ----------------------------------------

    if "Customer_Type" in df.columns and "Total_Sales" in df.columns:

        print("\n" + "=" * 50)
        print("SALES BY CUSTOMER TYPE")
        print("=" * 50)

        sales_by_customer = (
            df.groupby("Customer_Type")["Total_Sales"]
            .sum()
            .sort_values(ascending=False)
        )

        display(sales_by_customer)

        plt.figure(figsize=(7, 5))
        sales_by_customer.plot(kind="bar")
        plt.title("Sales by Customer Type")
        plt.xlabel("Customer Type")
        plt.ylabel("Total Sales")
        plt.xticks(rotation=0)
        plt.tight_layout()
        plt.show()

    # ----------------------------------------
    # 10. Sales by Gender
    # ----------------------------------------

    if "Gender" in df.columns and "Total_Sales" in df.columns:

        print("\n" + "=" * 50)
        print("SALES BY GENDER")
        print("=" * 50)

        sales_by_gender = (
            df.groupby("Gender")["Total_Sales"]
            .sum()
            .sort_values(ascending=False)
        )

        display(sales_by_gender)

        plt.figure(figsize=(7, 5))
        sales_by_gender.plot(kind="bar")
        plt.title("Sales by Gender")
        plt.xlabel("Gender")
        plt.ylabel("Total Sales")
        plt.xticks(rotation=0)
        plt.tight_layout()
        plt.show()

    # ----------------------------------------
    # 11. Top Products
    # ----------------------------------------

    if "Product" in df.columns and "Total_Sales" in df.columns:

        print("\n" + "=" * 50)
        print("SALES BY PRODUCT")
        print("=" * 50)

        sales_by_product = (
            df.groupby("Product")["Total_Sales"]
            .sum()
            .sort_values(ascending=False)
        )

        display(sales_by_product)

        plt.figure(figsize=(9, 5))
        sales_by_product.plot(kind="bar")
        plt.title("Sales by Product")
        plt.xlabel("Product")
        plt.ylabel("Total Sales")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    # ----------------------------------------
    # 12. Correlation heatmap
    # ----------------------------------------

    numeric_df = df.select_dtypes(include=np.number)

    if not numeric_df.empty and numeric_df.shape[1] >= 2:

        print("\n" + "=" * 50)
        print("CORRELATION HEATMAP")
        print("=" * 50)

        plt.figure(figsize=(8, 6))
        sns.heatmap(
            numeric_df.corr(),
            annot=True,
            cmap="coolwarm",
            fmt=".2f"
        )
        plt.title("Correlation Heatmap")
        plt.tight_layout()
        plt.show()

    print("\n" + "=" * 50)
    print("✅ EDA COMPLETED SUCCESSFULLY")
    print("=" * 50)
```
